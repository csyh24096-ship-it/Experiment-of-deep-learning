#②モデルの生成

```python
weights = ResNet50_Weights.IMAGENET1K_V2
resnet = resnet50(weights=weights)
resnet.eval()
preprocess = weights.transforms()
print('preprocess:\n',preprocess)
```

#⑧前処理およびバッチ化

```python
batch = preprocess(image).unsqueeze(0)
print(batch.shape)
print('batch.type', type(batch))
```

#⑨ResNet50による画像の推論

```python
with torch.no_grad():
  prediction = resnet(batch).squeeze(0).softmax(0)
print(prediction.shape)
print('output type', type(prediction))
```

#⑩ResNet50の推論結果の出コード

```python
k = 5
topK = torch.topk(prediction,k)
print('predict:')
for i in range(k):
  print(f"{weights.meta['categories'][topK.indices[i]]}:{100 * topK.values[i]:.1f}%")
```
